## Palindrome number

```python
# plaindrome number using string
text=input('enter a string')
if text==text[::-1]:
    print('palindrome')
else:
    print('not plaindrome')
    
```

```python
# plaindrome string using while loop
text = input('Enter a string')
left = 0
right = len(text)-1
is_palindrome = True
while left < right:
    if text[left] != text[right]:
        is_palindrome = false
        break
    left = left + 1
    right = right - 1
if is_palindrome:
     print('palindrome')
else:
     print('not palindrome')
        
        
```
